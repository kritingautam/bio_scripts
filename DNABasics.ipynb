```python
# Nucleotide Count: Open the text file and read the sequence
with open("sequence.txt", "r") as file:
    dna_sequence = file.read().strip()

# Count each individual nucleotide using the built-in .count() method
a_count = dna_sequence.count('A')
t_count = dna_sequence.count('T')
c_count = dna_sequence.count('C')
g_count = dna_sequence.count('G')

# Print out the results using formatted strings
print(f"Adenine (A): {a_count}")
print(f"Thymine (T): {t_count}")
print(f"Cytosine (C): {c_count}")
print(f"Guanine (G): {g_count}")
```

```python
#dna_to_rna: Open the text file and read the sequence
with open("sequence.txt", "r") as file:
    dna_sequence = file.read().strip()

# Initialize an empty string to build our RNA sequence
rna_sequence = ""

# Loop through every single letter (nucleotide) in the DNA string
for nucleotide in dna_sequence:
    if nucleotide == 'T':
        rna_sequence += 'U'  # Replace T with U
    else:
        rna_sequence += nucleotide  # Keep A, C, G the same

# Print the final transcribed RNA sequence
print("RNA Sequence:", rna_sequence)
```

```python
# Open the text file and read the sequence
with open("sequence.txt", "r") as file:
    dna_sequence = file.read().strip()

# Calculate total length
total_length = len(dna_sequence)

# Count G and C characters
g_count = dna_sequence.count('G')
c_count = dna_sequence.count('C')

# Calculate GC percentage
gc_percentage = ((g_count + c_count) / total_length) * 100

# Print the result rounded to 2 decimal places
print(f"GC Content: {gc_percentage:.2f}%")
```

```python
# Reversen Complement: Open the text file and read the sequence
with open("sequence.txt", "r") as file:
    dna_sequence = file.read().strip()

# Define complement mapping using a dictionary
complement_dict = {'A': 'T', 'T': 'A', 'C': 'G', 'G': 'C'}

# Create the complement string by looping through the sequence backwards
reverse_complement = ""
for base in reversed(dna_sequence):
    reverse_complement += complement_dict.get(base, base)  # default to base if unknown

# Print the results
print("Original DNA:         ", dna_sequence)
print("Reverse Complement:   ", reverse_complement)
```

```python
# Open the FASTA file
filename = "sequence.fasta"

header = ""
sequence_lines = []

with open(filename, "r") as file:
    for line in file:
        # Remove trailing newlines/spaces
        line = line.strip()
        
        # Check if the line is a FASTA header
        if line.startswith(">"):
            header = line
        else:
            # Append sequence lines to our list
            sequence_lines.append(line)

# Join all individual sequence fragments into a single continuous string
full_sequence = "".join(sequence_lines)

# Print the results
print("Header found:  ", header)
print("Clean Sequence:", full_sequence)
print("Total Length:  ", len(full_sequence))
```
