```python
import numpy as np
import pandas as pd

df = pd.read_csv("gene_expression.csv")
expression_array = df['expression_level'].to_numpy()

print("--- NumPy Expression Statistics ---")
print(f"Mean Expression:   {np.mean(expression_array):.2f}")
print(f"Median Expression: {np.median(expression_array):.2f}")
print(f"Standard Dev:      {np.std(expression_array):.2f}")
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("gene_expression.csv")
plt.figure(figsize=(8, 5))

colors = ['tab:red' if cond == 'treated' else 'tab:blue' for cond in df['condition']]
plt.bar(df['gene_id'], df['expression_level'], color=colors, alpha=0.85, edgecolor='black')

plt.xlabel("Gene ID", fontweight='bold')
plt.ylabel("Expression Level", fontweight='bold')
plt.title("Gene Expression Levels: Treated vs. Control", fontweight='bold')
plt.grid(axis='y', linestyle='--', alpha=0.5)

plt.savefig("gene_expression_plot.png", dpi=300, bbox_inches='tight')
print("Plot successfully generated and saved as 'gene_expression_plot.png'")
plt.show()
```
