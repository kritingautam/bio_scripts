```python
import pandas as pd

df = pd.read_csv("gene_expression.csv")
print("--- Original Data ---")
print(df)

high_expression = df[df['expression_level'] > 2.0]
print("\n--- Filtered Data (Expression > 2.0) ---")
print(high_expression)

high_expression.to_csv("high_expression_genes.csv", index=False)
print("\nFiltered data saved to 'high_expression_genes.csv'")
```
